# Dataset 1 —— Appliances Energy Prediction (UCI)

> **Situação:** A empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

### Etapa A — Orange Data Mining

Nesta etapa (realizada no Orange Data Mining) foram executados os seguintes procedimentos:

1. Carregue o arquivo energydata_complete.csv no Orange Data Mining utilizando o widget File.
2. Inspecione os registros no Data Table e identifique os atributos de consumo, temperatura e
umidade.
3. Utilize Select Columns para manter Appliances, lights, pelo menos três atributos de
temperatura e três de umidade.
4. Verifique se existem valores ausentes. Este dataset é disponibilizado pelo UCI sem valores//
ausentes, mas a verificação deve fazer parte do procedimento.
5. Utilize Data Sampler para gerar uma amostra aleatória de 10% dos registros.
6. Exporte a amostra em CSV utilizando Save Data.

### Etapa B — Python / Pandas

1. Carregue a amostra no Pandas e apresente head(), shape, info() e describe().
2. Renomeie Appliances para Consumo_Eletrodomesticos e simplifique o nome de pelo menos
três atributos ambientais.
3. Determine o maior consumo de eletrodomésticos registrado na amostra.
4. Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame contendo
somente os registros acima desse limite.
5. Conte os registros selecionados e calcule qual percentual da amostra eles representam.
6. Calcule a temperatura média de T1 e crie outro DataFrame contendo simultaneamente
consumo acima de 70% do máximo e temperatura acima da média.
7. Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo
critério.

```python
df_original = pd.read_csv('/content/energydata_complete.csv')

colunas_selecionadas = ['Appliances', 'lights', 'T1', 'T2', 'T3', 'RH_1', 'RH_2', 'RH_3']
df_selecionado = df_original[colunas_selecionadas]

df_limpo = df_selecionado.dropna()

df = df_limpo.sample(frac=0.10, random_state=42)

print("=== Head ===")
display(df.head())
print(f"\nShape: {df.shape}")
print("\n=== Info ===")
df.info()
print("\n=== Describe ===")
display(df.describe())

df = df.rename(columns={
    'Appliances': 'Consumo_Eletrodomesticos',
    'T1': 'Temp_Cozinha',
    'RH_1': 'Umid_Cozinha',
    'T2': 'Temp_Sala'
})

max_consumo = df['Consumo_Eletrodomesticos'].max()
print(f"\nMaior consumo registrado: {max_consumo}")

limiar_70 = max_consumo * 0.70
df_acima_70 = df[df['Consumo_Eletrodomesticos'] > limiar_70].copy()

qtd_acima_70 = len(df_acima_70)
perc_acima_70 = (qtd_acima_70 / len(df)) * 100
print(f"Registros acima de 70% do consumo máximo: {qtd_acima_70} ({perc_acima_70:.2f}%)")

media_temp_t1 = df['Temp_Cozinha'].mean()
df_critico = df[
    (df['Consumo_Eletrodomesticos'] > limiar_70) &
    (df['Temp_Cozinha'] > media_temp_t1)
].copy()

qtd_critico = len(df_critico)
perc_critico = (qtd_critico / len(df)) * 100
print(f"Temperatura média T1 (Cozinha): {media_temp_t1:.2f}°C")
print(f"Registros críticos (Consumo > 70% E Temp > média): {qtd_critico} ({perc_critico:.2f}%)")

print("\n=== Conclusão ===")
print("A inclusão da temperatura como segundo critério diminui a quantidade de registros observados. "
      "Isso indica que nem todo pico de consumo acontece quando a residência está mais quente que a média. "
      "Cruzar essas informações ajuda a isolar os momentos em que o calor externo ou interno pode estar "
      "forçando equipamentos (como refrigeradores ou ar-condicionado) a operarem em capacidade máxima.")
```

# Dataset 2 — Steel Industry Energy Consumption (UCI)

> **Situação:** A equipe de gestão de energia de uma indústria siderúrgica deseja localizar situações de consumo elevado e verificar se esses períodos coincidem com condições de carga elevada ou valores menos favoráveis de fator de potência.

### Etapa A — Orange Data Mining

Nesta etapa (realizada no Orange Data Mining) foram executados os seguintes procedimentos:

1. Carregamento do arquivo `Steel_industry_data.csv` via widget **File**;
2. Inspeção dos registros no **Data Table**;
3. Seleção das colunas com **Select Columns**: `Usage_kWh`, variáveis de potência reativa (`Lagging_Current_Reactive.Power_kVarh`, `Leading_Current_Reactive_Power_kVarh`), fatores de potência (`Lagging_Current_Power_Factor`, `Leading_Current_Power_Factor`), `WeekStatus`, `Day_of_week` e `Load_Type`;
4. Verificação de valores ausentes — nenhum encontrado neste dataset;
5. Geração de **amostra aleatória de 20%** via **Data Sampler** (random state fixo para reprodutibilidade);
6. Exportação da amostra em CSV via **Save Data**.

### Etapa B — Python / Pandas

```python
df_orig2 = pd.read_csv('/content/Steel_industry_data.csv')
cols2 = [
    'Usage_kWh', 'Lagging_Current_Reactive.Power_kVarh', 'Leading_Current_Reactive_Power_kVarh',
    'Lagging_Current_Power_Factor', 'Leading_Current_Power_Factor', 'WeekStatus', 'Day_of_week', 'Load_Type'
]
df_amostra2 = df_orig2[cols2].sample(frac=0.20, random_state=42)
```

```python

df2 = df_amostra2.copy()

df2 = df2.rename(columns={
    'Usage_kWh'                           : 'Consumo_kWh',
    'Lagging_Current_Power_Factor'         : 'FP_Defasado',
    'Leading_Current_Power_Factor'         : 'FP_Adiantado',
    'Lagging_Current_Reactive.Power_kVarh' : 'Pot_Reativa_Def_kVarh',
    'Leading_Current_Reactive_Power_kVarh' : 'Pot_Reativa_Adi_kVarh'
})

print('Colunas após renomeação:', df2.columns.tolist())
```

```python

print('=== head() ===')
display(df2.head())

print('\n=== shape ===')
print(f'Linhas: {df2.shape[0]} | Colunas: {df2.shape[1]}')

print('\n=== info() ===')
df2.info()

print('\n=== describe() ===')
display(df2.describe())
```

```python

max_consumo = df2['Consumo_kWh'].max()
limiar_75   = 0.75 * max_consumo

print(f'Maior consumo registrado : {max_consumo:.3f} kWh')
print(f'Limiar de 75% do máximo  : {limiar_75:.3f} kWh')
```

```python

df2_alto = df2[df2['Consumo_kWh'] > limiar_75].copy()

qtd_alto  = len(df2_alto)
perc_alto = qtd_alto / len(df2) * 100

print(f'Registros acima de 75% do máximo : {qtd_alto}')
print(f'Percentual da amostra            : {perc_alto:.2f}%')
```

```python

print('Categorias em Load_Type:', df2['Load_Type'].unique())

df2_max_load = df2_alto[df2_alto['Load_Type'] == 'Maximum_Load']
qtd_max_load = len(df2_max_load)

print(f'\nRegistros de consumo elevado com Load_Type = Maximum_Load : {qtd_max_load}')
print(f'Percentual sobre os registros de consumo elevado : {qtd_max_load / qtd_alto * 100:.2f}%')
```

```python
print('Estatísticas de FP_Defasado:')
print(df2_alto['FP_Defasado'].describe())

limite_fp = 90.0
```

```python
df2_critico = df2[
    (df2['Consumo_kWh'] > limiar_75) &
    (df2['FP_Defasado'] < limite_fp)
].copy()

qtd_critico  = len(df2_critico)
perc_critico = qtd_critico / len(df2) * 100

print(f'Registros críticos (consumo > 75% máx E FP_Def < {limite_fp}) : {qtd_critico}')
print(f'Percentual da amostra : {perc_critico:.2f}%')
display(df2_critico.head())
```

### B8. Interpretação — por que esses registros merecem atenção?

O primeiro filtro (consumo > 75% do máximo) identifica os momentos de maior demanda de energia ativa, ou seja, quando a usina está trabalhando próximo da sua capacidade máxima. Isso já é um alerta operacional relevante.

O segundo critério (FP Defasado < 0,90) acrescenta um sinal de **ineficiência elétrica**: um fator de potência baixo significa que a rede está circulando uma corrente reativa elevada sem contribuir para o trabalho útil da máquina. Isso gera:

- **Perdas ôhmicas maiores** nos cabos e transformadores (calor desnecessário);
- **Risco de penalização tarifária** — concessionárias cobram multa quando o FP cai abaixo de 0,92;
- **Sobrecarga dos equipamentos** em um momento já crítico de consumo.

A combinação dos dois critérios, portanto, seleciona os registros onde a usina consome muito **e** ao mesmo tempo opera de forma ineficiente. São os intervalos onde uma ação corretiva (banco de capacitores, ajuste de carga) teria o maior impacto positivo.

# Dataset 3 — Power Consumption of Tetouan City (UCI)

> **Situação:** Identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais nos momentos de maior demanda.

### Etapa A — Orange Data Mining

1. Carregamento do arquivo `Tetuan City power consumption.csv` via widget **File**;
2. Inspeção dos registros no **Data Table**;
3. Seleção com **Select Columns**: `Zone 1 Power Consumption`, `Zone 2  Power Consumption`, `Zone 3  Power Consumption`, `Temperature`, `Humidity`, `Wind Speed`;
4. Verificação de valores ausentes — nenhum encontrado;
5. **Amostra aleatória de 15%** via **Data Sampler**;
6. Exportação em CSV.

### Etapa B — Python / Pandas

```python
df3 = pd.read_csv('/content/Tetuan City power consumption.csv')

rename_map = {}
for col in df3.columns:
    col_strip = col.strip()
    if 'Zone 1' in col_strip:
        rename_map[col] = 'Consumo_Zona_1'
    elif 'Zone 2' in col_strip:
        rename_map[col] = 'Consumo_Zona_2'
    elif 'Zone 3' in col_strip:
        rename_map[col] = 'Consumo_Zona_3'
    elif 'Temperature' in col_strip:
        rename_map[col] = 'Temperatura'
    elif 'Humidity' in col_strip:
        rename_map[col] = 'Umidade'
    elif 'Wind Speed' in col_strip:
        rename_map[col] = 'Vento'

df3 = df3.rename(columns=rename_map)
print('Colunas após renomeação:', df3.columns.tolist())
```

```python
print('=== head() ===')
display(df3.head())

print('\n=== shape ===')
print(f'Linhas: {df3.shape[0]} | Colunas: {df3.shape[1]}')

print('\n=== info() ===')
df3.info()

print('\n=== describe() ===')
display(df3.describe())
```

```python
max_z1 = df3['Consumo_Zona_1'].max()
max_z2 = df3['Consumo_Zona_2'].max()
max_z3 = df3['Consumo_Zona_3'].max()

print(f'Máximo Zona 1: {max_z1:.2f}')
print(f'Máximo Zona 2: {max_z2:.2f}')
print(f'Máximo Zona 3: {max_z3:.2f}')
```

```python
maximos = {
    'Consumo_Zona_1': max_z1,
    'Consumo_Zona_2': max_z2,
    'Consumo_Zona_3': max_z3
}
zona_pico = max(maximos, key=maximos.get)
valor_pico = maximos[zona_pico]

print(f'Zona com maior pico de consumo: {zona_pico} ({valor_pico:.2f})')
```

```python
limiar_zona = 0.70 * valor_pico
print(f'Limiar de 70% do máximo ({zona_pico}): {limiar_zona:.2f}')

df3_alta_demanda = df3[df3[zona_pico] > limiar_zona].copy()
```

```python
qtd_alta = len(df3_alta_demanda)
perc_alta = qtd_alta / len(df3) * 100

print(f'Registros acima de 70% do máximo ({zona_pico}): {qtd_alta}')
print(f'Percentual da amostra: {perc_alta:.2f}%')
```

```python
media_temp = df3['Temperatura'].mean()
print(f'Temperatura média da amostra: {media_temp:.2f} °C')

df3_critico = df3[
    (df3[zona_pico] > limiar_zona) &
    (df3['Temperatura'] > media_temp)
].copy()

qtd_crit  = len(df3_critico)
perc_crit = qtd_crit / len(df3) * 100

print(f'Registros críticos (consumo > 70% máx E temperatura > média): {qtd_crit}')
print(f'Percentual da amostra: {perc_crit:.2f}%')
```

### B7. Comparação e interpretação

| Critério | Registros | % da amostra |
|---|---|---|
| Consumo > 70% do máximo (1 condição) | `qtd_alta` | `perc_alta`% |
| + Temperatura acima da média (2 condições) | `qtd_crit` | `perc_crit`% |

> **O que ocorreu com a quantidade de registros?**
>
> Ao incluir a condição de temperatura, o conjunto de registros foi **reduzido**, pois agora apenas os períodos que satisfazem **simultaneamente** os dois critérios permanecem. Isso revela que nem todos os picos de consumo ocorrem sob altas temperaturas — parte da demanda elevada pode ser explicada por outros fatores, como períodos de pico industrial ou concentração de atividade em determinados horários, independentemente do calor. O segundo DataFrame representa os momentos de maior risco térmico-elétrico combinado, onde o resfriamento (ar-condicionado, ventilação) provavelmente intensifica a demanda da rede.

# Dataset 5 — Wind & Solar Energy Production (Kaggle)

> **Situação:** Comparar a ocorrência de períodos de alta produção solar e alta produção eólica utilizando limites relativos às suas escalas próprias.

### Etapa A — Orange Data Mining

1. Carregamento do dataset via widget **File**;
2. Inspeção dos atributos disponíveis no **Data Table**;
3. Seleção com **Select Columns**: atributos temporais (`Date`, `Start_Hour`, `End_Hour`, `Day_of_Year`, `Day_Name`, `Month_Name`, `Season`) e variáveis de produção (`Source`, `Production`);
4. Observação dos valores mínimos e máximos de produção para cada fonte;
5. Verificação de valores ausentes;
6. **Amostra aleatória de 20%** via **Data Sampler**;
7. Exportação em CSV.

### Etapa B — Python / Pandas

```python
df_orig5 = pd.read_csv('/content/Energy Production Dataset.csv')
cols5 = ['Date', 'Start_Hour', 'End_Hour', 'Day_of_Year', 'Day_Name', 'Month_Name', 'Season', 'Source', 'Production']
df_amostra5 = df_orig5[cols5].sample(frac=0.20, random_state=42)
```

```python
df5 = df_amostra5.copy()

print(f'Registros na amostra: {df5.shape[0]}')
print(f'Colunas: {df5.columns.tolist()}')
print(f'\nValores ausentes: {df5.isnull().sum().sum()}')
if df5.isnull().sum().sum() > 0:
    df5 = df5.dropna()
    print('-> Registros com ausentes removidos.')
```

```python
solar  = df5[df5['Source'] == 'Solar'].copy()
eolica = df5[df5['Source'] == 'Wind'].copy()

solar.rename(columns={'Production': 'Geracao_Solar'}, inplace=True)
eolica.rename(columns={'Production': 'Geracao_Eolica'}, inplace=True)

print(f'Registros solares na amostra : {len(solar)}')
print(f'Registros eólicos na amostra : {len(eolica)}')
```

```python
print('=== head() — Solar ===')
display(solar.head())

print('\n=== describe() — Solar ===')
display(solar[['Geracao_Solar']].describe())

print('\n=== describe() — Eólica ===')
display(eolica[['Geracao_Eolica']].describe())
```

```python
max_solar  = solar['Geracao_Solar'].max()
max_eolica = eolica['Geracao_Eolica'].max()

print(f'Máximo de Geração Solar  : {max_solar:.2f} MWh')
print(f'Máximo de Geração Eólica : {max_eolica:.2f} MWh')
```

```python
limiar_solar  = 0.70 * max_solar
limiar_eolica = 0.70 * max_eolica

print(f'Limiar 70% Solar  : {limiar_solar:.2f} MWh')
print(f'Limiar 70% Eólica : {limiar_eolica:.2f} MWh')
```

```python
df5_alta_solar  = solar[solar['Geracao_Solar']   > limiar_solar]
df5_alta_eolica = eolica[eolica['Geracao_Eolica'] > limiar_eolica]

qtd_solar  = len(df5_alta_solar)
qtd_eolica = len(df5_alta_eolica)

perc_solar  = qtd_solar  / len(solar)  * 100
perc_eolica = qtd_eolica / len(eolica) * 100

print(f'Alta geração Solar  (>70% máx) : {qtd_solar}  registros ({perc_solar:.2f}% dos registros solares)')
print(f'Alta geração Eólica (>70% máx) : {qtd_eolica} registros ({perc_eolica:.2f}% dos registros eólicos)')
```

```python
if perc_solar > perc_eolica:
    mais_freq = 'Solar'
    menos_freq = 'Eólica'
    diff = perc_solar - perc_eolica
else:
    mais_freq = 'Eólica'
    menos_freq = 'Solar'
    diff = perc_eolica - perc_solar

print(f'A fonte {mais_freq} aparece com maior frequência acima de 70% do seu próprio máximo.')
print(f'Diferença percentual entre as fontes: {diff:.2f} pontos percentuais.')
```

### B7. Por que não usar o mesmo valor numérico como limite para as duas fontes?

Usinas solares e eólicas possuem **escalas de produção completamente diferentes** dependendo da capacidade instalada, localização geográfica e tecnologia. Se definíssemos um corte absoluto (por exemplo, 5.000 MWh) como limiar de "alta geração" para ambas:

- A fonte com maior capacidade poderia **sempre** estar acima do limite — tornando o filtro inútil para ela;
- A fonte com menor capacidade poderia **nunca** atingi-lo — excluindo-a completamente da análise.

Ao calcular **70% do máximo de cada variável separadamente**, normalizamos as escalas: estamos avaliando quando cada parque opera próximo do **teto da sua própria capacidade**, independentemente do valor absoluto. Isso garante uma comparação justa entre fontes de natureza e porte distintos.

# Dataset 6 — Individual Household Electric Power Consumption (UCI)

> **Situação:** Identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio.

### Etapa A — Orange Data Mining

1. Carregamento do arquivo **original** `household_power_consumption.txt` (separador `;`, valor nulo `?`) via widget **File**;
2. Inspeção no **Data Table** e verificação dos registros com valores ausentes;
3. Tratamento dos dados ausentes conforme procedimento trabalhado em aula;
4. Seleção com **Select Columns**: `Global_active_power`, `Global_reactive_power`, `Voltage`, `Global_intensity`, `Sub_metering_1`, `Sub_metering_2`, `Sub_metering_3`;
5. **Amostra aleatória de 10%** dos dados tratados via **Data Sampler**;
6. Exportação em CSV.

### Etapa B — Python / Pandas

```python
df_orig6 = pd.read_csv('/content/household_power_consumption.txt', sep=';', na_values='?', low_memory=False)
cols6 = ['Global_active_power', 'Global_reactive_power', 'Voltage', 'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']

df_amostra6 = df_orig6[cols6].dropna().sample(frac=0.10, random_state=42)
```

```python
df6 = df_amostra6.copy()

df6 = df6.rename(columns={
    'Global_active_power'   : 'pot_ativa',
    'Global_reactive_power' : 'pot_reativa',
    'Voltage'               : 'tensao',
    'Global_intensity'      : 'corrente',
    'Sub_metering_1'        : 'sub_1',
    'Sub_metering_2'        : 'sub_2',
    'Sub_metering_3'        : 'sub_3'
})

for col in df6.columns:
    df6[col] = pd.to_numeric(df6[col], errors='coerce')

df6 = df6.dropna()
print(f'Amostra carregada: {df6.shape[0]} registros | {df6.shape[1]} colunas')
print('Colunas:', df6.columns.tolist())
```

```python
print('=== head() ===')
display(df6.head())

print('\n=== shape ===')
print(f'Linhas: {df6.shape[0]} | Colunas: {df6.shape[1]}')

print('\n=== info() ===')
df6.info()

print('\n=== describe() ===')
display(df6.describe())
```

```python
max_pot  = df6['pot_ativa'].max()
limiar_75 = 0.75 * max_pot

print(f'Máximo de potência ativa (amostra) : {max_pot:.3f} kW')
print(f'Limiar de 75% do máximo            : {limiar_75:.3f} kW')
```

```python
df6_pot_alta = df6[df6['pot_ativa'] > limiar_75].copy()

qtd_pot  = len(df6_pot_alta)
perc_pot = qtd_pot / len(df6) * 100

print(f'Registros com potência ativa > 75% do máximo : {qtd_pot}')
print(f'Percentual da amostra                        : {perc_pot:.4f}%')
```

```python
media_corrente = df6['corrente'].mean()
print(f'Corrente média da amostra: {media_corrente:.3f} A')

df6_critico = df6[
    (df6['pot_ativa'] > limiar_75) &
    (df6['corrente']  > media_corrente)
].copy()

qtd_crit  = len(df6_critico)
perc_crit = qtd_crit / len(df6) * 100

print(f'\nRegistros críticos (potência > 75% E corrente > média) : {qtd_crit}')
print(f'Percentual da amostra                                   : {perc_crit:.4f}%')
```

```python
reducao = (1 - qtd_crit / qtd_pot) * 100 if qtd_pot > 0 else 0

print(f'Registros — 1 condição (só potência)   : {qtd_pot}')
print(f'Registros — 2 condições (pot + corrente): {qtd_crit}')
print(f'Redução ao adicionar a corrente          : {reducao:.1f}%')
display(df6_critico.head())
```

### B7. Efeito da inclusão da corrente como segunda condição

| Critério | Registros | % da amostra |
|---|---|---|
| Potência ativa > 75% do máximo (1 condição) | `qtd_pot` | `perc_pot`% |
| + Corrente acima da média (2 condições) | `qtd_crit` | `perc_crit`% |

Ao adicionar a condição de corrente, o conjunto foi **reduzido e refinado**. O primeiro filtro captura os momentos de maior demanda de energia ativa, mas inclui situações onde o circuito pode estar sob alta potência sem necessariamente sobrecarregar fisicamente a infraestrutura elétrica.

A corrente (intensidade) é o fator que determina o desgaste real dos condutores, disjuntores e tomadas. Quando a potência ativa é alta **e simultaneamente** a corrente está acima do comportamento médio da residência, temos um "estresse genuíno da rede": o calor gerado nos cabos é proporcional a I², o que aumenta o risco de falha. O segundo DataFrame isola exatamente esses momentos, eliminando os "falsos alertas" onde a potência é alta mas a corrente permanece em patamar normal.
